## Task 1: Brown Corpus Text Preprocessing & Lexical Normalization

**Objective**:
1. Load and segment raw text from the NLTK Brown Corpus into distinct sentences.
2. Construct a preprocessing pipeline that lowercases tokens, removes non-alphabetic noise, and filters out standard English stopwords.
3. Compare raw versus cleaned representations across sample sentences.
4. Compute key corpus statistics (lexical diversity, compression ratio).

```python
import nltk
import string
import pandas as pd
from nltk.corpus import brown, stopwords
from nltk.tokenize import sent_tokenize, word_tokenize

# Ensure dependencies are available
for pkg in ['brown', 'punkt', 'punkt_tab', 'stopwords']:
    nltk.download(pkg, quiet=True)

print("Corpora and tokenizers ready.")
```

### 1. Corpus Ingestion & Sentence Extraction

```python
# Extract raw text from Brown corpus
raw_brown_text = brown.raw()
segmented_sentences = sent_tokenize(raw_brown_text)[:20]

print(f"Extracted first {len(segmented_sentences)} sample sentences from Brown corpus.")
for i in range(3):
    print(f"\nSample {i+1}: {segmented_sentences[i].strip()}")
```

### 2. Preprocessing & Stopword Removal Pipeline

```python
import re

english_stopwords = set(stopwords.words("english"))

def preprocess_sentence(sentence: str, stopword_filter: set) -> list[str]:
    # Strips POS tags if present, converts to lower case, tokenizes,
    # and filters out non-alphabetic tokens and common stopwords.
    clean_text = re.sub(r"/[a-z0-9\$\-]+", "", sentence, flags=re.IGNORECASE)
    tokens = word_tokenize(clean_text.lower())
    clean_tokens = [tok for tok in tokens if tok.isalpha() and tok not in stopword_filter]
    return clean_tokens

cleaned_sentence_list = [preprocess_sentence(s, english_stopwords) for s in segmented_sentences]

# Tabular comparison for first 5 sentences
comparison_records = []
for idx in range(5):
    comparison_records.append({
        "Index": idx + 1,
        "Original Sentence": segmented_sentences[idx].strip(),
        "Cleaned Tokens": ", ".join(cleaned_sentence_list[idx]),
        "Original Word Count": len(segmented_sentences[idx].split()),
        "Cleaned Token Count": len(cleaned_sentence_list[idx])
    })

comparison_table = pd.DataFrame(comparison_records)
display(comparison_table)
```

### 3. Quantitative Corpus Insights

```python
total_raw_words = sum(len(s.split()) for s in segmented_sentences)
total_cleaned_tokens = sum(len(toks) for toks in cleaned_sentence_list)
unique_cleaned_vocab = len(set(tok for toks in cleaned_sentence_list for tok in toks))

print("=" * 50)
print("LEXICAL STATISTICS (First 20 Sentences)")
print("=" * 50)
print(f"Raw Word Count        : {total_raw_words}")
print(f"Cleaned Token Count   : {total_cleaned_tokens}")
print(f"Unique Vocabulary     : {unique_cleaned_vocab}")
print(f"Stopword/Punct Filter : {((total_raw_words - total_cleaned_tokens) / total_raw_words) * 100:.1f}% reduction")
print(f"Lexical Diversity     : {unique_cleaned_vocab / total_cleaned_tokens:.3f}")
```
