## Task 2: Comparative Evaluation of Stemming vs Lemmatization

**Objective**:
1. Sample vocabulary from the Brown Corpus and integrate required benchmark terms (`playing`, `studies`, `running`, `better`, `cars`).
2. Apply the heuristic `PorterStemmer` and vocabulary-based `WordNetLemmatizer`.
3. Construct a comparative DataFrame highlighting morphological differences between both techniques.

```python
import nltk
import pandas as pd
from nltk.corpus import brown
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Ensure resources are downloaded
for pkg in ['brown', 'wordnet', 'omw-1.4']:
    nltk.download(pkg, quiet=True)

# Extract alphabetic vocabulary from Brown
corpus_words = [w.lower() for w in brown.words() if w.isalpha()]
required_benchmarks = ['playing', 'studies', 'running', 'better', 'cars']

# Select 30 representative words + the 5 required benchmark words
selected_test_words = corpus_words[:30] + required_benchmarks
print(f"Total test vocabulary size: {len(selected_test_words)}")
```

### Morphological Processing & Comparison

```python
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

results_list = []
for term in selected_test_words:
    stem_val = stemmer.stem(term)
    lemma_val = lemmatizer.lemmatize(term)
    
    results_list.append({
        "Original Word": term,
        "Porter Stem": stem_val,
        "WordNet Lemma": lemma_val,
        "Identical": stem_val == lemma_val
    })

comparison_df = pd.DataFrame(results_list)
print("Complete Vocabulary Comparison (First 15 Rows):")
display(comparison_df.head(15))
```

### Evaluation of Required Target Words

```python
# Highlight the 5 required benchmark words
benchmark_subset = comparison_df[comparison_df["Original Word"].isin(required_benchmarks)].reset_index(drop=True)

print("=" * 60)
print("REQUIRED BENCHMARK WORDS EVALUATION")
print("=" * 60)
display(benchmark_subset)
```

### Observations:
- **`playing` / `running`**: Porter Stemmer correctly isolates base stems `play` and `run`. WordNet Lemmatizer defaults to noun interpretation unless POS is supplied.
- **`studies`**: Stemmer reduces to `studi` (a non-dictionary stem), whereas Lemmatizer morphologically recovers `study`.
- **`better`**: Lemmatizer preserves `better` unless specified as adjective with comparative degree.
- **`cars`**: Both stemmer and lemmatizer successfully reduce plural noun `cars` to singular `car`.
