## Task 4: N-gram Language Modeling & Frequency Distribution

**Objective**:
1. Extract and normalize tokens from the Brown corpus.
2. Construct Unigram ($N=1$), Bigram ($N=2$), and Trigram ($N=3$) language models using `nltk.util.ngrams`.
3. Compute and analyze the Top 10 most frequent N-grams across each order.

```python
import nltk
from nltk.corpus import brown
from nltk.util import ngrams
from collections import Counter
import pandas as pd

nltk.download('brown', quiet=True)

# Extract lowercase alphabetic tokens
brown_tokens = [w.lower() for w in brown.words() if w.isalpha()]

print(f"Total alphabetic tokens extracted: {len(brown_tokens):,}")
print(f"First 15 tokens: {brown_tokens[:15]}")
```

### 1. Generating N-gram Sequences

```python
unigram_model = list(ngrams(brown_tokens, 1))
bigram_model  = list(ngrams(brown_tokens, 2))
trigram_model = list(ngrams(brown_tokens, 3))

unigram_dist = Counter(unigram_model)
bigram_dist  = Counter(bigram_model)
trigram_dist = Counter(trigram_model)

print(f"Generated Unigram count : {len(unigram_model):,}")
print(f"Generated Bigram count  : {len(bigram_model):,}")
print(f"Generated Trigram count : {len(trigram_model):,}")
```

### 2. Top 10 N-gram Frequency Rankings

```python
def format_ngram_table(counter: Counter, n_val: int):
    rows = []
    for gram, freq in counter.most_common(10):
        rows.append({
            f"{n_val}-gram": " ".join(gram),
            "Frequency": freq
        })
    return pd.DataFrame(rows)

print("Top 10 Unigrams:")
display(format_ngram_table(unigram_dist, 1))

print("\nTop 10 Bigrams:")
display(format_ngram_table(bigram_dist, 2))

print("\nTop 10 Trigrams:")
display(format_ngram_table(trigram_dist, 3))
```
