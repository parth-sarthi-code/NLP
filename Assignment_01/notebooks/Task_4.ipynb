```python
import nltk
from nltk.corpus import brown
from nltk.util import ngrams
from collections import Counter

nltk.download('brown')

# Get Brown Corpus words
words = brown.words()

# Retain lowercase alphabetic tokens
tokens = [word.lower() for word in words if word.isalpha()]

print("Total tokens:", len(tokens))
print("First 20 tokens:")
print(tokens[:20])
```

```python
# Generate N-grams
unigrams = list(ngrams(tokens, 1))
bigrams = list(ngrams(tokens, 2))
trigrams = list(ngrams(tokens, 3))

# Calculate frequencies
unigram_freq = Counter(unigrams)
bigram_freq = Counter(bigrams)
trigram_freq = Counter(trigrams)

print("Number of unigrams:", len(unigrams))
print("Number of bigrams:", len(bigrams))
print("Number of trigrams:", len(trigrams))
```

```python
print("Top 10 Unigrams:")
for gram, frequency in unigram_freq.most_common(10):
    print(gram, ":", frequency)

print("\nTop 10 Bigrams:")
for gram, frequency in bigram_freq.most_common(10):
    print(gram, ":", frequency)

print("\nTop 10 Trigrams:")
for gram, frequency in trigram_freq.most_common(10):
    print(gram, ":", frequency)
```
