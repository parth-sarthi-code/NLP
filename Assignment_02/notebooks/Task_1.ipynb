# Natural Language Processing Lab - Assignment 02
## Task 1: WordNet Lexical Database Exploration & Synset Analysis

**Objective**:
1. Select 10 diverse English words and query Princeton WordNet for their synset mappings.
2. Extract definitions, part-of-speech categories, and example usages.
3. Conduct polysemy analysis on words exhibiting multiple distinct semantic senses.

```python
import nltk
import pandas as pd
from nltk.corpus import wordnet as wn

nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)

print("WordNet semantic database loaded.")
```

### 1. Target Lexicon Selection

```python
target_words = [
    "bank", "play", "book", "light", "crane",
    "train", "mouse", "plant", "rock", "spring"
]

print("Selected target words:")
for w in target_words:
    syn_count = len(wn.synsets(w))
    print(f"- {w:<10} : {syn_count} synset(s)")
```

### 2. Detailed Synset Extraction

```python
lexicon_records = []

for word in target_words:
    synsets = wn.synsets(word)
    for syn in synsets:
        example_str = syn.examples()[0] if syn.examples() else "N/A"
        lexicon_records.append({
            "Target Word": word,
            "Synset Name": syn.name(),
            "POS": syn.pos(),
            "Definition": syn.definition(),
            "Example Usage": example_str
        })

lexicon_df = pd.DataFrame(lexicon_records)
print(f"Total synsets analyzed: {len(lexicon_df)}")
display(lexicon_df.head(15))
```

### 3. Polysemy Case Study: Multi-Sense Words

```python
polysemous_sample = ["bank", "light", "crane"]

for word in polysemous_sample:
    print(f"\n{'='*70}\nPOLYSEMY ANALYSIS: '{word.upper()}' ({len(wn.synsets(word))} Senses)\n{'='*70}")
    for idx, syn in enumerate(wn.synsets(word)[:4], 1):
        print(f"{idx}. [{syn.pos().upper()}] {syn.name()}: {syn.definition()}")
        if syn.examples():
            print(f'   Example: "{syn.examples()[0]}"')
```
