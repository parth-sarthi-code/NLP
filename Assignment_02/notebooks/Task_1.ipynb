```python

import nltk
import pandas as pd

# Download the WordNet corpus
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.corpus import wordnet as wn

print("WordNet loaded successfully!")
```

```python
# Select 10 English words for WordNet exploration

words = [
    "car",
    "bank",
    "bat",
    "dog",
    "light",
    "plant",
    "book",
    "teacher",
    "student",
    "computer"
]

print("Selected words:")
print(words)
print(f"\nTotal words: {len(words)}")
```

```python
# Display synsets for each selected word

for word in words:
    synsets = wn.synsets(word)
    
    print(f"\n{'=' * 60}")
    print(f"Word: {word}")
    print(f"Number of synsets: {len(synsets)}")
    
    for synset in synsets:
        print(f"  - {synset.name()}")
```

```python
# Display synset name, definition, and example sentences

for word in words:
    print(f"\n{'=' * 80}")
    print(f"WORD: {word.upper()}")
    print(f"{'=' * 80}")
    
    synsets = wn.synsets(word)
    
    for synset in synsets:
        print(f"\nSynset: {synset.name()}")
        print(f"Definition: {synset.definition()}")
        
        examples = synset.examples()
        
        if examples:
            print("Example:")
            for example in examples:
                print(f"  - {example}")
        else:
            print("Example: No example available")
```

```python
# Store WordNet information in a Pandas DataFrame

synset_data = []

for word in words:
    for synset in wn.synsets(word):
        examples = synset.examples()
        
        synset_data.append({
            "Word": word,
            "Synset": synset.name(),
            "Definition": synset.definition(),
            "Example": examples[0] if examples else "No example available"
        })

synset_df = pd.DataFrame(synset_data)

print("WordNet Synset Information:")
display(synset_df)
```

```python
# Identify words that have multiple synsets

multiple_synset_words = []

for word in words:
    synsets = wn.synsets(word)
    
    if len(synsets) > 1:
        multiple_synset_words.append(word)

print("Words with multiple synsets:")
for word in multiple_synset_words:
    print(f"- {word}: {len(wn.synsets(word))} synsets")
```

```python
# Analyze different meanings of three words with multiple synsets

selected_multiple_words = ["bank", "bat", "light"]

for word in selected_multiple_words:
    print(f"\n{'=' * 80}")
    print(f"Different meanings of: {word.upper()}")
    print(f"{'=' * 80}")
    
    for synset in wn.synsets(word):
        print(f"\nSynset: {synset.name()}")
        print(f"Definition: {synset.definition()}")
        
        if synset.examples():
            print(f"Example: {synset.examples()[0]}")
        else:
            print("Example: No example available")
```

## Observations

WordNet shows that a single English word can have multiple synsets,
with each synset representing a different meaning or sense.

For example:

- **Bank** can refer to a financial institution or the side of a river.
- **Bat** can refer to a flying mammal or a piece of sports equipment.
- **Light** can refer to illumination or something that is not heavy.

Therefore, the meaning of a word depends on the sense represented by
its synset and, in natural language, the context in which the word is used.
