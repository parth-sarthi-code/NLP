```python
# Task 4: Homonyms and Contextual Meaning

import nltk
import pandas as pd

# Download required NLTK resources
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.corpus import wordnet as wn

print("WordNet loaded successfully!")
```

```python
# Select five homonyms for contextual analysis

homonyms = [
    "bank",
    "bat",
    "bark",
    "light",
    "match"
]

print("Selected homonyms:")
for word in homonyms:
    print(f"- {word}")

print(f"\nTotal homonyms: {len(homonyms)}")
```

```python
# Create two contextual sentences for each homonym

context_examples = {
    "bank": [
        {
            "sentence": "She deposited her salary at the bank.",
            "meaning": "Financial institution",
            "clue": "deposited her salary"
        },
        {
            "sentence": "We sat on the bank of the river.",
            "meaning": "Side of a river",
            "clue": "river"
        }
    ],
    
    "bat": [
        {
            "sentence": "The bat flew out of the cave at night.",
            "meaning": "Flying mammal",
            "clue": "flew out of the cave"
        },
        {
            "sentence": "The baseball player hit the ball with a bat.",
            "meaning": "Sports equipment",
            "clue": "baseball player and hit the ball"
        }
    ],
    
    "bark": [
        {
            "sentence": "The dog began to bark loudly.",
            "meaning": "Sound made by a dog",
            "clue": "dog"
        },
        {
            "sentence": "The bark of the tree was rough.",
            "meaning": "Outer covering of a tree",
            "clue": "tree"
        }
    ],
    
    "light": [
        {
            "sentence": "Please turn on the light in the room.",
            "meaning": "Illumination",
            "clue": "turn on and room"
        },
        {
            "sentence": "This bag is very light and easy to carry.",
            "meaning": "Not heavy",
            "clue": "easy to carry"
        }
    ],
    
    "match": [
        {
            "sentence": "He struck a match to light the candle.",
            "meaning": "Small stick used to produce fire",
            "clue": "struck and light the candle"
        },
        {
            "sentence": "The football match starts at six o'clock.",
            "meaning": "Sports competition",
            "clue": "football"
        }
    ]
}

print("Contextual sentences created successfully!")
```

```python
# Display the contextual sentences for each homonym

for word, examples in context_examples.items():
    print(f"\n{'=' * 70}")
    print(f"WORD: {word.upper()}")
    print(f"{'=' * 70}")
    
    for i, example in enumerate(examples, start=1):
        print(f"\nSentence {i}: {example['sentence']}")
        print(f"Intended Meaning: {example['meaning']}")
        print(f"Contextual Clue: {example['clue']}")
```

```python
# Display WordNet synsets for each homonym

for word in homonyms:
    print(f"\n{'=' * 70}")
    print(f"WORD: {word.upper()}")
    print(f"{'=' * 70}")
    
    synsets = wn.synsets(word)
    
    for synset in synsets:
        print(f"\nSynset: {synset.name()}")
        print(f"Definition: {synset.definition()}")
        
        if synset.examples():
            print(f"Example: {synset.examples()[0]}")
        else:
            print("Example: No example available")
```

```python
# Display relevant WordNet synsets for the intended meanings

selected_synsets = {
    "bank": ["bank.n.01", "bank.n.01"],
    "bat": ["bat.n.01", "bat.n.02"],
    "bark": ["bark.n.01", "bark.n.02"],
    "light": ["light.n.01", "light.a.01"],
    "match": ["match.n.01", "match.n.01"]
}

for word, synset_names in selected_synsets.items():
    print(f"\n{word.upper()}")
    
    for synset_name in synset_names:
        synset = wn.synset(synset_name)
        print(f"\nSynset: {synset.name()}")
        print(f"Definition: {synset.definition()}")
```

```python
# Create the final contextual meaning DataFrame

context_data = []

for word, examples in context_examples.items():
    for example in examples:
        context_data.append({
            "Word": word,
            "Sentence": example["sentence"],
            "Intended Meaning": example["meaning"],
            "Contextual Clue": example["clue"]
        })

context_df = pd.DataFrame(context_data)

print("Final Contextual Meaning Table:")
display(context_df)
```

```python
# Analyze how surrounding words determine the intended meaning

context_analysis = [
    {
        "Word": "bank",
        "Context": "She deposited her salary at the bank.",
        "Important Context": "deposited her salary",
        "Explanation": "The words 'deposited her salary' indicate that bank refers to a financial institution."
    },
    {
        "Word": "bat",
        "Context": "The baseball player hit the ball with a bat.",
        "Important Context": "baseball player and hit the ball",
        "Explanation": "The baseball context indicates that bat refers to sports equipment."
    },
    {
        "Word": "bark",
        "Context": "The bark of the tree was rough.",
        "Important Context": "tree",
        "Explanation": "The word 'tree' indicates that bark refers to the outer covering of a tree."
    },
    {
        "Word": "light",
        "Context": "This bag is very light and easy to carry.",
        "Important Context": "easy to carry",
        "Explanation": "The phrase 'easy to carry' indicates that light means not heavy."
    },
    {
        "Word": "match",
        "Context": "The football match starts at six o'clock.",
        "Important Context": "football",
        "Explanation": "The word 'football' indicates that match refers to a sports competition."
    }
]

analysis_df = pd.DataFrame(context_analysis)

display(analysis_df)
```

## Observations

Context plays an important role in determining the intended meaning of a
word when the word has multiple meanings.

- **Bank:** Words such as "deposited" and "salary" indicate a financial
  institution, while "river" indicates the side of a river.
- **Bat:** Words related to baseball indicate sports equipment, while
  "flew" and "cave" indicate the flying mammal.
- **Bark:** The presence of a "dog" indicates the sound made by a dog,
  while "tree" indicates the outer covering of a tree.
- **Light:** "Turn on" and "room" indicate illumination, while "easy to
  carry" indicates that the object is not heavy.
- **Match:** "Struck" and "candle" indicate a fire-producing stick, while
  "football" indicates a sports competition.

Therefore, surrounding words provide important contextual clues that help
identify the intended meaning of a word.
