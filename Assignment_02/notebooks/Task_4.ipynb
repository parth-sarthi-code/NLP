## Task 4: Homonyms, Polysemy & Context-Driven Sense Disambiguation

**Objective**:
1. Select 5 polysemous homonyms: `bank`, `bat`, `bark`, `light`, `match`.
2. Construct distinct contextual sentences demonstrating different semantic senses for each word.
3. Align each sentence with its corresponding Princeton WordNet synset and evaluate context clues.

```python
import nltk
import pandas as pd
from nltk.corpus import wordnet as wn

nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)

# Define the 5 target homonyms
homonyms_list = ["bank", "bat", "bark", "light", "match"]
```

### 1. Contextual Sentences & Semantic Annotations

```python
disambiguation_dataset = [
    {
        "Word": "bank",
        "Sentence": "The entrepreneur deposited the venture capital funds at the local commercial bank.",
        "Intended Sense": "Financial Institution",
        "Context Clues": "deposited, funds, commercial, venture capital",
        "Target Synset": "depository_financial_institution.n.01"
    },
    {
        "Word": "bank",
        "Sentence": "We set up our fishing rods along the grassy bank of the winding river.",
        "Intended Sense": "Sloping river margin",
        "Context Clues": "fishing rods, grassy, winding river",
        "Target Synset": "bank.n.01"
    },
    {
        "Word": "bat",
        "Sentence": "The nocturnal fruit bat navigated the deep cave using echolocation.",
        "Intended Sense": "Flying Mammal (Chiroptera)",
        "Context Clues": "nocturnal, fruit, cave, echolocation",
        "Target Synset": "bat.n.01"
    },
    {
        "Word": "bat",
        "Sentence": "The baseball player purchased a lightweight aluminum bat for the championship.",
        "Intended Sense": "Sports Equipment / Club",
        "Context Clues": "baseball player, lightweight aluminum, championship",
        "Target Synset": "bat.n.05"
    },
    {
        "Word": "bark",
        "Sentence": "The guard dog started to bark frantically when a stranger approached.",
        "Intended Sense": "Canine Vocalization",
        "Context Clues": "guard dog, frantically, stranger approached",
        "Target Synset": "bark.n.04"
    },
    {
        "Word": "bark",
        "Sentence": "The rough outer bark of the ancient redwood tree protects it against parasites.",
        "Intended Sense": "Outer Covering of Tree",
        "Context Clues": "rough outer, ancient redwood tree, parasites",
        "Target Synset": "bark.n.01"
    },
    {
        "Word": "light",
        "Sentence": "Please flick the switch to turn on the ceiling light in the conference room.",
        "Intended Sense": "Illumination / Lamp",
        "Context Clues": "switch, turn on, ceiling, conference room",
        "Target Synset": "light.n.02"
    },
    {
        "Word": "light",
        "Sentence": "This modern carbon-fiber laptop is remarkably light and portable for travel.",
        "Intended Sense": "Low Weight / Not Heavy",
        "Context Clues": "carbon-fiber, portable, travel, remarkably",
        "Target Synset": "light.a.01"
    },
    {
        "Word": "match",
        "Sentence": "He struck the wooden match against the sandpaper strip to ignite the campfire.",
        "Intended Sense": "Ignition Stick for Fire",
        "Context Clues": "struck, wooden, sandpaper strip, ignite campfire",
        "Target Synset": "match.n.01"
    },
    {
        "Word": "match",
        "Sentence": "Thousands of passionate fans gathered at the stadium for the football match.",
        "Intended Sense": "Athletic Competition / Game",
        "Context Clues": "passionate fans, stadium, football",
        "Target Synset": "match.n.02"
    }
]

# Enrich records with WordNet synset definitions
for item in disambiguation_dataset:
    try:
        syn = wn.synset(item["Target Synset"])
        item["Synset Definition"] = syn.definition()
    except Exception:
        # Fallback to first matching synset
        syn = wn.synsets(item["Word"])[0]
        item["Target Synset"] = syn.name()
        item["Synset Definition"] = syn.definition()

context_df = pd.DataFrame(disambiguation_dataset)
display(context_df[["Word", "Sentence", "Intended Sense", "Context Clues", "Target Synset", "Synset Definition"]])
```
