```python
# Task 1: Corpus Analysis and N-gram Statistics

import nltk
from nltk.corpus import brown
from nltk.util import ngrams
from collections import Counter
import pandas as pd

# Download required NLTK resources
nltk.download('brown')
nltk.download('punkt')
nltk.download('punkt_tab')

print("Resources loaded successfully!")
```

```python
# (a) Load the Brown Corpus and select the news category
raw_words = brown.words(categories='news')

print(f"Total words in news category: {len(raw_words)}")
```

```python
# (b) Extract at least 5,000 tokens and perform basic preprocessing
# Preprocessing: convert to lowercase and retain alphabetic tokens
tokens = [word.lower() for word in raw_words if word.isalpha()][:5000]

print(f"Extracted preprocessed tokens: {len(tokens)}")
print("\nSample first 20 tokens:")
print(tokens[:20])
```

```python
# (c) Calculate total tokens and vocabulary size
total_tokens = len(tokens)
vocabulary = set(tokens)
vocab_size = len(vocabulary)

print(f"Total Tokens: {total_tokens}")
print(f"Vocabulary Size (Unique Words): {vocab_size}")
print(f"Lexical Diversity (Vocab / Total): {vocab_size / total_tokens:.4f}")
```

```python
# (d) Generate unigram, bigram, and trigram sequences
unigrams = tokens
bigrams = list(ngrams(tokens, 2))
trigrams = list(ngrams(tokens, 3))

unigram_counts = Counter(unigrams)
bigram_counts = Counter(bigrams)
trigram_counts = Counter(trigrams)

print(f"Total Unigrams: {len(unigrams)}")
print(f"Total Bigrams: {len(bigrams)}")
print(f"Total Trigrams: {len(trigrams)}")
```

```python
# (e) Display the ten most frequent examples of each

print("Top 10 Most Frequent Unigrams:")
print("-" * 35)
for word, count in unigram_counts.most_common(10):
    print(f"{word:<15} : {count}")

print("\nTop 10 Most Frequent Bigrams:")
print("-" * 35)
for gram, count in bigram_counts.most_common(10):
    print(f"{' '.join(gram):<25} : {count}")

print("\nTop 10 Most Frequent Trigrams:")
print("-" * 35)
for gram, count in trigram_counts.most_common(10):
    print(f"{' '.join(gram):<35} : {count}")
```

```python
# Summary table of top 10 N-grams
top_ngrams_df = pd.DataFrame({
    "Top Unigrams": [f"{w} ({c})" for w, c in unigram_counts.most_common(10)],
    "Top Bigrams": [f"{' '.join(g)} ({c})" for g, c in bigram_counts.most_common(10)],
    "Top Trigrams": [f"{' '.join(g)} ({c})" for g, c in trigram_counts.most_common(10)]
})

display(top_ngrams_df)
```

## Observations

### (f) How a Language Corpus Supports Statistical Language Analysis

A standard language corpus such as the Brown Corpus serves as the empirical foundation for statistical language analysis:

1. **Probability Estimation**: Statistical language models estimate word and transition probabilities directly from corpus frequency distributions rather than handcrafted rules:
   $$P(w_n \mid w_{n-1}) = \frac{\text{Count}(w_{n-1}, w_n)}{\text{Count}(w_{n-1})}$$

2. **Capturing Syntactic and Collocational Regularities**: Higher-order N-grams (bigrams and trigrams) capture natural idioms and syntactic structures (such as *"of the"*, *"in the"*, *"for the first time"*), showing which word sequences commonly co-occur in natural usage.

3. **Zipfian Distribution & Lexical Sparsity**: The corpus demonstrates that a small number of grammatical function words account for a large portion of occurrences, while most words occur infrequently. This insight guides smoothing techniques and vocabulary design in NLP pipelines.

4. **Domain-Specific Modeling**: By selecting specific genres (such as *news*), corpora allow models to learn vocabulary, sentence structures, and conventions representative of target communicative domains.
