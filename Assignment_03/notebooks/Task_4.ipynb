```python
# (d) Load sci.space and comp.graphics from 20 Newsgroups
space_data = fetch_20newsgroups(subset='train', categories=['sci.space'], remove=('headers', 'footers', 'quotes'))
graphics_data = fetch_20newsgroups(subset='train', categories=['comp.graphics'], remove=('headers', 'footers', 'quotes'))

print(f"Total sci.space posts:     {len(space_data.data)}")
print(f"Total comp.graphics posts: {len(graphics_data.data)}")
```

```python
# (e) Extract and display at least 20 frequent domain-specific terms
stop_words = set(stopwords.words('english')).union({
    'writes', 'article', 'would', 'one', 'like', 'get', 'also', 'know', 'think',
    'use', 'could', 'may', 'people', 'say', 'make', 'time', 'good', 'see', 'two',
    'well', 'even', 'first', 'way', 'much', 'new', 'want', 'need', 'system', 'edu'
})

def get_top_terms(text_list, top_n=20):
    words = []
    for doc in text_list:
        tokens = re.findall(r'[a-zA-Z]{3,}', doc.lower())
        words.extend([w for w in tokens if w not in stop_words])
    return Counter(words).most_common(top_n)

top_space = get_top_terms(space_data.data, 20)
top_graphics = get_top_terms(graphics_data.data, 20)

domain_df = pd.DataFrame({
    "sci.space Top Terms": [f"{w} ({c})" for w, c in top_space],
    "comp.graphics Top Terms": [f"{w} ({c})" for w, c in top_graphics]
})

display(domain_df)
```

## Observations

### (f) How Ambiguity, Coreference, and Domain Vocabulary Affect NLP Performance

1. **Lexical Ambiguity:** Words with multiple meanings (polysemy or homonymy) present challenges for NLP pipelines. If a model lacks sufficient surrounding context, it may pick the wrong sense (e.g., misinterpreting financial *"bank"* as a river bank in translation or semantic search).
2. **Coreference Resolution:** Pronouns (such as *"she"*, *"it"*, *"they"*) point back to previously mentioned entities. If a model fails to resolve which antecedent a pronoun refers to, downstream tasks like question answering and summarization will attribute actions or facts to the wrong entity.
3. **Domain-Specific Vocabulary:** NLP models trained on general text (like news or Wikipedia) degrade when applied to specialized domains like space science or computer graphics. Specialized terminology (such as *"shuttle"*, *"orbit"*, *"lunar"*, *"jpeg"*, *"polygon"*) either appear as out-of-vocabulary words or have different frequency distributions, requiring domain adaptation.
